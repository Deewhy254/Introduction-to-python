```python
import random
# TIC TAC GAME
board = [" "] * 9
#player info
player1 = input("enter player  name X: ")
player2 = input("enter player  name O: ")
score_x = 0
score_o = 0

# Display board
def display_board():
   print("\n")
   print(board[0] + "|" + board[1] + "|" + board[2])
   print("--+--+--")
   print(board[3] + "|" + board[4] + "|" + board[5])
   print("--+--+--")
   print(board[6] + "|" + board[7] + "|" + board[8])
   print("\n")

#Display Scoreboard
def display_scoreboard():
    print("SCORE BOARD")
    print(player1, "(X)", score_x)
    print(player2, "(0)", score_o)
    print("---------------------")

#check the winner
def check_winner(player):
    win_positions = [
        (0,1,2),(3,4,5),(6,7,8),
        (0,3,6),(1,4,7),(2,5,8),
        (0,4,8),(2,4,6)
    ]
    for pos in win_positions:
        if board[pos[0]] == board[pos[1]] == board[pos[2]] == player:
            return True
    return False

# check draw
def check_draw():
    return " " not in board
# real game
def play_game():
    global score_x, score_o
    current_player = "X"
    while True:
        display_scoreboard()
        display_board()

        try:
          move = int(input(f"{player1 if current_player == 'X' else player2} ({current_player}), choose position (1-9): "))-1
          if move < 0 or move > 8:
            print("Invalid position. Try to choose between 1 and 9.")
            continue
          if board[move] != " ":
            print("Position already taken. Try again.")
            continue
          board[move] = current_player
        except ValueError:
          print("please enter valid number.")
          continue
        #check winner
        if check_winner(current_player):
          display_board()

          if current_player == "X":
            print(f"Congratulations {player1}! You won!")
            score_x += 1
          else:
            print(f"Congratulations {player2}! You won!")
            score_o += 1
          break

      # check draw
        if check_draw():
            display_board()
            print("It's a draw!")
            break
      # switch player
        current_player = "O" if current_player == "X" else "X"

#replay the game
while True:
  board = [" "] * 9
  play_game()
  again = input("Do you want to play again? (yes/no): ")
  if again.lower() != "yes":
    print("\nFINAL SCOREBOARD:")
    display_scoreboard()
    print("Thank you so much for coming it's nice to have played our game tic tac toe!")
    break
```
